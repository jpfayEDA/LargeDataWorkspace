# Access Box Data 
This notebook demonstrates how to use large datasets stored on a box drive in your versioned workspace. For this to work, you'll need the BoxDrive software installed on your machine, and you will need to be logged in to your Box account via the BoxDrive app. 

The path to the box folder on your machine depends on the operating system used (Windows or MacOS). Here, we use the `platform` package to determine what os is used and then creates a Path object pointing to the appropriate path. From there, we can add folders and files to the path object to access specific files that exist in the Box drive. 

In this example, we'll read the `CTD_chem_gene_ixns.csv` file that lives in the class Box drive (`ENVIRON 712L-Students`), within the `Data`/`CTD-Jalal` folder, loading its contents into a dataframe. 

```python
#Import packages
import platform
from pathlib import Path
import pandas as pd
```

```python
#Set a path to the main box folder based on machine's OS
if platform.system() == 'Windows':
    box_path = Path.home() / 'Box'
else:
    box_path = Path.home() / "Library" / "CloudStorage" / "Box-Box"

#Confirm that the box path is valid
if box_path.exists():
    print(f'Box folder found at {box_path}')
else:
    print(f'Failed to locate the Box folder.\nIs the BoxDrive app running?\nAre you logged in?')
```

```python
#Create a Path object to the a specific folder on the Box drive
the_file = box_path / 'ENVIRON 712L-Students' / 'Data' / 'CTD-Jalal' / 'CTD_chem_gene_ixns.csv'
print(the_file.exists())
```

```python
#Read a file (first 10 records) into a dataframe
df = pd.read_csv(the_file,nrows=10)
df.head()
```
